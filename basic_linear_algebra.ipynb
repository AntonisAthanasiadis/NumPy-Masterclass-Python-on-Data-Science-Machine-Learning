## Matrix Operations & Solving Systems

Key concepts: Determinant, Inverse, Eigenvalues/Eigenvectors, and Solving Linear Systems ($Ax = b$).

```python
import numpy as np

# Define a 2x2 square matrix
M = np.array([
    [4, 7],
    [2, 6]
])

# 1. Determinant
det = np.linalg.det(M)
print(f"Determinant of M: {det:.2f}")

# 2. Matrix Inverse
M_inv = np.linalg.inv(M)
print("\nInverse of M:\n", M_inv)
print("Verify M @ M_inv (Identity):\n", np.round(M @ M_inv))

# 3. Eigenvalues and Eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(M)
print("\nEigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)

# 4. Solving a System of Linear Equations (Ax = b)
# Equations: 2x + 3y = 8,  4x + 9y = 22
A_eq = np.array([[2, 3], [4, 9]])
b_eq = np.array([8, 22])
x = np.linalg.solve(A_eq, b_eq)
print("\nSolution to system Ax = b (x, y):", x)
```
