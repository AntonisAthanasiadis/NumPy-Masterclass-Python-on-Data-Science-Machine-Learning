## 1. Binary Files (`.npy` and `.npz`)

- `np.save` / `np.load`: Save and load a single array/matrix.
- `np.savez` / `np.load`: Save multiple matrices into a single compressed/uncompressed archive.

```python
import numpy as np

# Create sample matrix
matrix_a = np.array([[1, 2, 3], [4, 5, 6]])
matrix_b = np.eye(3)

# Save single array (.npy format)
np.save('matrix_a.npy', matrix_a)
loaded_a = np.load('matrix_a.npy')
print("Loaded Matrix A from .npy:\n", loaded_a)

# Save multiple arrays into a zipped archive (.npz format)
np.savez('matrices_archive.npz', mat1=matrix_a, mat2=matrix_b)
loaded_archive = np.load('matrices_archive.npz')
print("\nKeys in archived .npz:", loaded_archive.files)
print("Loaded mat1 from archive:\n", loaded_archive['mat1'])
```

## 2. Text and CSV Files (`.csv` / `.txt`)

- `np.savetxt`: Export matrix data to plain text or CSV format.
- `np.loadtxt` / `np.genfromtxt`: Read matrix data from plain text or CSV files.

```python
# Save matrix to a CSV file
data_matrix = np.array([
    [10.5, 20.2, 30.8],
    [40.1, 50.6, 60.3]
])

np.savetxt('data.csv', data_matrix, delimiter=',', header='Col1,Col2,Col3', fmt='%.2f')
print("Saved matrix to 'data.csv'")

# Load matrix back from CSV
loaded_csv = np.loadtxt('data.csv', delimiter=',', skiprows=1)
print("\nLoaded Matrix from CSV:\n", loaded_csv)
```
