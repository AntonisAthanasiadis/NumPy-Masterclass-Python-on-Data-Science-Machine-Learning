## 1. Universal Functions (ufuncs) Overview
Ufuncs operate on `ndarray` objects in an element-by-element fashion, executing optimized, compiled C code under the hood.


```python
import numpy as np
import matplotlib.pyplot as plt

# Create sample vector
x = np.array([0, np.pi/6, np.pi/4, np.pi/3, np.pi/2])

sin_x = np.sin(x)
cos_x = np.cos(x)

exp_x = np.exp(x)
log_x = np.log(x + 1)  # offset to avoid log(0)

print("x (radians):", x)
print("sin(x):     ", np.round(sin_x, 4))
print("cos(x):     ", np.round(cos_x, 4))
print("exp(x):     ", np.round(exp_x, 4))
print("log(x+1):   ", np.round(log_x, 4))
```

## 2. Visualizing 2D Vectors & Vector Addition
We can visualize 2D vectors and vector addition using Matplotlib's `quiver` function.


```python
# Define two vectors
v1 = np.array([3, 2])
v2 = np.array([1, 4])
v_sum = v1 + v2

plt.figure(figsize=(7, 7))
plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.8)

# Plot vector 1
plt.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='blue', label=f'v1 {v1}')

# Plot vector 2 (originating at head of v1 to show addition visually)
plt.quiver(v1[0], v1[1], v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='orange', label=f'v2 {v2}')

# Plot resultant vector (v1 + v2)
plt.quiver(0, 0, v_sum[0], v_sum[1], angles='xy', scale_units='xy', scale=1, color='green', label=f'v1 + v2 {v_sum}')

plt.xlim(-1, 6)
plt.ylim(-1, 7)
plt.grid(True, linestyle=':', alpha=0.6)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Vector Addition Visualization')
plt.legend(loc='upper left')
plt.show()
```

## 3. Plotting Continuous Functions using ufuncs
Ufuncs let us quickly compute output vectors over continuous domains created with `np.linspace`.


```python
x = np.linspace(-2 * np.pi, 2 * np.pi, 200)

# Evaluate trigonometric ufuncs
y_sin = np.sin(x)
y_cos = np.cos(x)
y_sinc = np.sinc(x / np.pi)  # normalized sinc function ufunc

plt.figure(figsize=(10, 5))
plt.plot(x, y_sin, label='sin(x)', color='tab:blue', linewidth=2)
plt.plot(x, y_cos, label='cos(x)', color='tab:orange', linewidth=2)
plt.plot(x, y_sinc, label='sinc(x)', color='tab:green', linestyle='--', linewidth=2)

plt.title('Trigonometric Universal Functions (ufuncs)')
plt.xlabel('x (radians)')
plt.ylabel('ufunc(x)')
plt.grid(True, linestyle=':', alpha=0.6)
plt.axhline(0, color='black', linewidth=0.8)
plt.legend()
plt.show()
```
