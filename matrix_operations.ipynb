## 1. Creating Matrices

We start by importing `numpy` and creating matrices manually from lists, as well as generating standard matrices like Zeros, Ones, and Identity matrices.

```python
import numpy as np

# Creating a 2x3 matrix from python lists
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Matrix A:")
print(A)
print("Shape of A:", A.shape)

# Creating common matrix types
zeros_mat = np.zeros((3, 3))
ones_mat = np.ones((2, 4))
identity_mat = np.eye(3)
random_mat = np.random.randint(1, 10, size=(3, 3))

print("\n3x3 Zeros Matrix:\n", zeros_mat)
print("\n3x3 Identity Matrix:\n", identity_mat)
print("\n3x3 Random Matrix:\n", random_mat)
```

## 2. Accessing and Modifying Elements

You can access and update specific elements, whole rows, or columns by index.

```python
# Define a 3x3 matrix M
M = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Original Matrix M:\n", M)

# Access individual element (Row index 1, Column index 2)
print("\nElement at row 1, col 2:", M[1, 2])

# Modify a single element
M[0, 0] = 99
print("\nAfter updating element (0,0) to 99:\n", M)

# Modify an entire row
M[1, :] = [44, 55, 66]
print("\nAfter updating Row 1:\n", M)

# Modify an entire column
M[:, 2] = [3, 6, 9]
print("\nAfter updating Column 2:\n", M)
```

## 3. Reshaping, Transposing, and Stacking

Transforming matrix shapes and combining multiple matrices together.

```python
# Transposing M
M_transposed = M.T
print("Transposed Matrix M:\n", M_transposed)

# Reshaping a 1D array into a 3x4 matrix
flat_array = np.arange(1, 13)
reshaped_mat = flat_array.reshape(3, 4)
print("\nReshaped 3x4 Matrix:\n", reshaped_mat)

# Vertical and Horizontal Stacking
mat1 = np.array([[1, 2], [3, 4]])
mat2 = np.array([[5, 6], [7, 8]])

v_stacked = np.vstack((mat1, mat2))
h_stacked = np.hstack((mat1, mat2))

print("\nVertically Stacked:\n", v_stacked)
print("\nHorizontally Stacked:\n", h_stacked)
```

## 4. Matrix Arithmetic

Demonstrating element-wise matrix arithmetic versus true matrix multiplication.

```python
X = np.array([[1, 2], [3, 4]])
Y = np.array([[5, 6], [7, 8]])

# Element-wise Addition
print("Element-wise Addition (X + Y):\n", X + Y)

# Element-wise Multiplication
print("\nElement-wise Multiplication (X * Y):\n", X * Y)

# True Matrix Multiplication (Dot product)
print("\nMatrix Multiplication (X @ Y):\n", X @ Y)
```
