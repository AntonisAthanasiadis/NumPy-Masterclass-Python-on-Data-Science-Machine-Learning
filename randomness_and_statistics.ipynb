## 1. Generating Random Vectors
NumPy makes it simple to generate random vectors filled with integers, uniform decimals, or values from a normal distribution.


```python
import numpy as np
import matplotlib.pyplot as plt

# Set random seed so results are reproducible
np.random.seed(1345)

# 1. Random integers (e.g., numbers between 1 and 100)
rand_ints = np.random.randint(1, 101, size=10)

# 2. Uniform random floats between 0 and 1
rand_floats = np.random.rand(5)

print("Random Integers:", rand_ints)
print("Random Floats:  ", np.round(rand_floats, 3))
```

## 2. Calculating Basic Statistics
Once you have a data vector, you can quickly calculate core summary statistics.


```python
# Create a sample vector of test scores
scores = np.array([78, 85, 92, 65, 70, 95, 88, 82, 90, 74])

mean_val = np.mean(scores)
median_val = np.median(scores)
std_dev = np.std(scores)
min_val = np.min(scores)
max_val = np.max(scores)

print("Scores:", scores)
print("-----------------------")
print("Mean (Average):", mean_val)
print("Median:        ", median_val)
print("Std Deviation: ", round(std_dev, 2))
print("Min Score:     ", min_val)
print("Max Score:     ", max_val)
```

## 3. Simple Probability Simulations
Vectors are great for running quick simulations, like flipping a coin or rolling a die thousands of times.


```python
# Simulate 1,000 dice rolls (values 1 through 6)
rolls = np.random.randint(1, 7, size=1000)

# Count occurrences of each face
unique, counts = np.unique(rolls, return_counts=True)

plt.figure(figsize=(7, 4))
plt.bar(unique, counts, color='coral', edgecolor='black', width=0.5)

plt.title('Simulating 1,000 Dice Rolls')
plt.xlabel('Dice Face')
plt.ylabel('Count')
plt.xticks(unique)
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.show()
```
