## 1. Matrix Broadcasting

Broadcasting allows arithmetic operations between arrays of different dimensions without unnecessary memory copying.

```python
import numpy as np

# Scalar Broadcasting
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print("Matrix + 10:\n", matrix + 10)

# Row Broadcasting (adding a 1D vector across rows)
row_vector = np.array([100, 200, 300])
print("\nMatrix + Row Vector:\n", matrix + row_vector)

# Column Broadcasting (adding a column vector across columns)
col_vector = np.array([[10], [20]])  # Shape (2, 1)
print("\nMatrix + Column Vector:\n", matrix + col_vector)
```

## 2. Advanced Indexing & Slicing

Accessing and modifying matrix elements using boolean masks and index arrays.

```python
A = np.array([
    [12, 25, 30],
    [45, 50, 61],
    [78, 82, 90]
])

# Boolean Filtering
mask = A > 50
print("Boolean Mask (A > 50):\n", mask)
print("Elements > 50:", A[mask])

# Conditional Modification
A[A > 50] = 0
print("\nMatrix after setting elements > 50 to 0:\n", A)

# Integer Array Indexing
B = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
row_idx = [0, 1, 2]
col_idx = [2, 1, 0]
print("\nSelected elements at (0,2), (1,1), (2,0):", B[row_idx, col_idx])
```
